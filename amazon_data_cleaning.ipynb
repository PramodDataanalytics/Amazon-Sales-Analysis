```python
import re
import pandas as pd

def clean(text):
    if pd.isna(text):
        return 'Unknown'

    text = str(text).lower().strip()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    if len(text) < 2:
        return 'Unknown'

    return text.title()
```

```python
amazon["State"]=amazon["State"].apply(clean)

```

```python
amazon.dropna(subset= ["City","State"],how="all")
state_map=amazon[["City","State"]].drop_duplicates(subset="City").set_index("City")["State"]
amazon["State"].fillna(amazon["City"].map(state_map))
```

```python
amazon["City"]=amazon["City"].apply(clean)
```

```python
# Drop rows where 'City' or 'State' is 'Unknown'
initial_rows = amazon.shape[0]
amazon= amazon[~((amazon['City'] == 'Unknown') | (amazon['State'] == 'Unknown'))]

print(f"Initial number of rows: {initial_rows}")
```
